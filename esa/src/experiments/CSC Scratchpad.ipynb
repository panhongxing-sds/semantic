### Load the questions and generate responses

```python
potato = read_potato_questions()
model_name = CFG["OAI"]["oai_llm_small"]
model = models.OAILLM(model=model_name)
fname = 'data/potato_responses_%s.json'%model_name
```

```python
idx = 0
data = {}
for query in list(potato):
    responses, log_probs = model.generate(
        prompt=query,
        n_generations=100
    )
    data[idx] = {
        'query': query,
        'responses': responses,
        'log_probs': log_probs
    }
    idx += 1
```

```python
save_dict_to_json(data, fname)
```
