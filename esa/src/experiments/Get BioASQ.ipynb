```python
import sys, os
import numpy as np
import pandas as pd
import json

from csc import *
from exp_utils import *

current_dir = os.path.dirname(os.path.abspath('__file__'))
parent_dir = os.path.dirname(current_dir)
sys.path.insert(0, parent_dir)
```

```python
models = [
    "gemma-2-9b-it",
    "Llama-3.1-8B-Instruct",
    "Mistral-7B-Instruct-v0.3",
    "Phi-3.5-mini-instruct",
]
```

```python
fname = f"{current_dir}/data/working/training11b.json"
with open(fname) as f:
    bioasq = json.load(f)
```

```python
bioasq_questions = bioasq["questions"]
print(len(bioasq_questions))
max_num_correct_answers = 0
for i in range(len(bioasq_questions)):
    x = len(bioasq_questions[i]["ideal_answer"])
    if x > max_num_correct_answers:
        max_num_correct_answers = x
print(max_num_correct_answers)
```

```python
reformatted_bioasq = {}
for i in range(len(bioasq_questions)):
    idx = str(i)
    reformatted_bioasq[idx] = {
        "query": bioasq_questions[i]["body"],
        "context": "",
        "true_ans": bioasq_questions[i]["ideal_answer"],
        "answerable": True,
        "responses": [],
        "log_probs": [],
        "cluster_ids": {},
        "single_response": "",
        "rating": ""
    }
```

```python
for p in ["preprompt", "no_preprompt"]:
    for model in models:
        fname = f"{current_dir}/data/{p}/{model}/bioasq_final_results.json"
        with open(fname, "w") as f:
            json.dump(reformatted_bioasq, f)
```
